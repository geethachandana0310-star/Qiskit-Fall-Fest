```python
from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator
from qiskit.visualization import plot_histogram
import matplotlib.pyplot as plt
```

```python
# Create a circuit with exactly 1 qubit and 1 classical bit
qc1 = QuantumCircuit(1, 1)

# Apply a rotation gate to create biased probabilities
qc1.ry(1.0472, 0)

# Measure the qubit
qc1.measure(0, 0)

# Display the circuit
print(qc1.draw())

# Execute the circuit for 100 shots
simulator = AerSimulator()
job1 = simulator.run(qc1, shots=100)
result1 = job1.result()

# Get measurement counts
counts1 = result1.get_counts()

print("Circuit 1 Results:", counts1)

# Display histogram
plot_histogram(counts1)
plt.show()
```

```python
# Create a circuit with 1 qubit and 1 classical bit
qc1 = QuantumCircuit(1, 1)

# Apply RY rotation gate
qc1.ry(3.14159 / 3, 0)

# Measure the qubit
qc1.measure(0, 0)

# Display the circuit
display(qc1.draw("mpl"))
```

```python
# Create simulator
simulator = AerSimulator()

# Execute circuit for 100 shots
job1 = simulator.run(qc1, shots=100)

# Get results
result1 = job1.result()
counts1 = result1.get_counts()

# Print results
print("Circuit 1 Results:", counts1)
```

```python
%matplotlib inline

import matplotlib.pyplot as plt
from qiskit.visualization import plot_histogram

print("Counts:", counts1)

fig = plot_histogram(
    counts1,
    title="Circuit 1: Zero Occurs More Frequently"
)

plt.show()
display(fig)
```

```python
# Create a circuit with 1 qubit and 1 classical bit
qc2 = QuantumCircuit(1, 1)

# Apply RY rotation gate
qc2.ry(2 * 3.14159 / 3, 0)

# Measure the qubit
qc2.measure(0, 0)

# Display the circuit
display(qc2.draw("mpl"))
```

```python
# Execute Circuit 2 for 100 shots
job2 = simulator.run(qc2, shots=100)

# Get results
result2 = job2.result()
counts2 = result2.get_counts()

# Print results
print("Circuit 2 Results:", counts2)
```

```python
plot_histogram(
    counts2,
    title="Circuit 2: One Occurs More Frequently"
)

plt.show()
```

```python
plot_histogram(
    [counts1, counts2],
    legend=["Circuit 1: 0-biased", "Circuit 2: 1-biased"],
    title="Comparison of Quantum Measurement Results"
)

plt.show()
```

Explanation

The RY rotation gate changes the quantum state and controls the measurement probabilities.

In Circuit 1, the RY(π/3) gate produces approximately 75% probability for 0 and 25% probability for 1.

In Circuit 2, the RY(2π/3) gate produces approximately 25% probability for 0 and 75% probability for 1.

Both circuits are executed for 100 shots. Their measurement results are displayed using histograms. Actual measurement counts may vary because quantum measurement outcomes are probabilistic.
